```python
import pandas as pd

# Chargement du fichier propre créé à l'exercice 08
df = pd.read_csv("../exercice-08/data/titanic_propre.csv")
```

```python
df.groupby(["class", "sex"], observed=False)["survived"].mean() * 100
```

Oui, les femmes présentent un taux de survie nettement supérieur à celui des hommes dans l'ensemble des classes (par exemple ~96 % contre ~37 % en 1re classe, et ~50 % contre ~13 % en 3e classe).

```python
# Découpage en tranches d'âge
tranches = [0, 12, 18, 60, 100]
labels = ["Enfant (0-12)", "Ado (12-18)", "Adulte (18-60)", "Senior (60+)"]

df["tranche_age"] = pd.cut(
    df["age"], bins=tranches, labels=labels, right=False
)
survie_age = df.groupby("tranche_age")["survived"].mean() * 100
print(survie_age)
```

Oui, les enfants (0-12 ans) ont bénéficié d'une priorité marquée lors de l'évacuation avec un taux de survie supérieur à celui des autres tranches d'âge.

```python
# Prix moyen par port
prix_par_port = df.groupby("embark_town")["fare"].mean()
print("Prix moyen par port :\n", prix_par_port)

# Répartition des classes par port (%)
repartition_classes = (
    pd.crosstab(df["embark_town"], df["class"], normalize="index")
    * 100
)
print("\nRépartition des classes (%):\n", repartition_classes)
```

Cherbourg affiche le tarif moyen le plus élevé car c'est le port où la proportion de passagers voyageant en 1ère classe était la plus forte.
