```python
import pandas as pd

# On charge d'abord le fichier nettoyé
titanic = pd.read_csv("../exercice-08/data/titanic_propre.csv")

# On crée le dictionnaire ville -> pays
ports = pd.DataFrame(
    {
        "embark_town": ["Southampton", "Cherbourg", "Queenstown"],
        "pays": ["Angleterre", "France", "Irlande"],
    }
)
```

```python
titanic_enrichi = pd.merge(titanic, ports, on="embark_town", how="left")
print("Nombre de lignes :", len(titanic_enrichi))
```

```python
# 1. Compter les passagers par pays
print(titanic_enrichi["pays"].value_counts())

# 2. Taux de survie moyen selon le pays d'embarquement
print(titanic_enrichi.groupby("pays")["survived"].mean() * 100)
```

```python
# On retire volontairement Queenstown de la table ports
ports_sans_q = ports[ports["embark_town"] != "Queenstown"]

# Test A : inner join
test_inner = pd.merge(titanic, ports_sans_q, on="embark_town", how="inner")
print("Lignes en inner :", len(test_inner))

# Test B : left join
test_left = pd.merge(titanic, ports_sans_q, on="embark_town", how="left")
print("Lignes en left :", len(test_left))
```

Le `inner join` ne garde que les correspondances parfaites et supprime les 77 passagers de Queenstown. 
Le `left join` conserve tous les passagers de la table principale et met simplement `NaN` dans la colonne pays pour ceux de Queenstown.
