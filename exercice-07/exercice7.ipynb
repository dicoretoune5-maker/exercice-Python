```python
import seaborn as sns
titanic = sns.load_dataset("titanic")
nombre = (titanic["age"] <18)
print(nombre)
```

```python
titanic.nlargest(5, "fare")[["fare", "pclass", "embark_town"]]
```

```python
premiere_classe = titanic[titanic["pclass"] == 1]
part_femmes = (premiere_classe["sex"] == "female").mean()
print(f"{part_femmes:.1%}")
```

```python
femme_1re = titanic[(titanic["sex"] == "female") & (titanic["pclass"] == 1)]
taux_femmes_1re = femme_1re["survived"].mean()
hommes_3e = titanic[(titanic["sex"] == "male") & (titanic["pclass"] == 3)]
taux_hommes_3e = hommes_3e["survived"].mean()
print(f"Taux de survie des femmes de femme de 1re classe : {taux_femmes_1re:.1%}")
print(f"Taux de survie des hommes de 3e classe : {taux_hommes_3e:.1%}")
```
